1、SSH连接的建立

建立SSH连接的第一步是创建一个SSH客户端对象。我们可以使用Paramiko的SSHClient类来创建一个SSH客户端对象并建立连接：

```python
import paramiko

client = paramiko.SSHClient()
client.load_system_host_keys()
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

#连接ssh服务器
client.connect('192.168.229.200', port=22, username='work', password='991227')
```

在这里，我们使用connect()方法连接SSH服务器，其中参数hostname是SSH服务器的主机名或IP地址，port是SSH服务器的端口号（通常是22），username是用于认证的用户名，password是用于认证的密码。

3. 执行远程命令

连接成功后，我们可以使用SSH客户端对象执行远程命令：

```python
import paramiko

client = paramiko.SSHClient()
client.load_system_host_keys() # 加载系统的host密钥
client.set_missing_host_key_policy(paramiko.AutoAddPolicy()) # 自动添加新的host密钥

#连接ssh服务器
client.connect('192.168.229.200', port=22, username='work', password='991227')

# 执行远程命令
stdin,stdout,stderr = client.exec_command('ls')
output = stdout.read().decode('utf-8')
print(output)
client.close()  # 关闭SSH连接
```

在这里，我们使用exec_command()方法执行远程命令，并将其结果保存在stdout中。注意，我们需要使用decode('utf-8')将输出解码为可读文本。

4. 上传和下载文件

使用Paramiko，我们还可以实现文件的上传和下载。下面是一个文件上传的示例：

```python
import paramiko

client = paramiko.SSHClient()
client.load_system_host_keys()
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

#连接ssh服务器
client.connect('192.168.229.200', port=22, username='work', password='991227')

sftp = client.open_sftp() #打开sftp会话
sftp.put('test.txt', '/home/work/test.txt') #上传文件
sftp.close()  # 关闭SFTP会话
client.close()  # 关闭SSH连接
```

在这里，open_sftp()方法打开一个SFTP会话，并使用put()方法将本地文件上传到远程服务器。

而文件的下载则可以使用以下代码：

```python
import paramiko

client = paramiko.SSHClient()
client.load_system_host_keys()
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

#连接ssh服务器
client.connect('192.168.229.200', port=22, username='work', password='991227')

sftp = client.open_sftp() #打开sftp会话
sftp.get('/home/work/test.txt', 'test.txt') #下载文件
sftp.close()  # 关闭SFTP会话
client.close()  # 关闭SSH连接
```
